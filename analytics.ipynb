# Анализ данных интернет-магазина

В данном блокноте выполняется подготовка, анализ и визуализация полученных данных. 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 1. Подготовка и очистка данных
Загружаем csv-файл, проверяем данные на пропуски, удаляем дубликаты, приводим типы данных, даты и время к корректному виду.

```python
df = pd.read_csv('data/ecommerce_analytics_dataset.csv')

df.columns = [str(col).strip().lower().replace(' ', '_') for col in df.columns]

df.isnull().sum()

df = df.drop_duplicates()

if 'order_date' in df.columns:
    df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

num_cols = ['quantity', 'unit_price', 'total_sales', 'shipping_cost', 'discount']
for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.dropna(subset=['total_sales'])

df.head()
```

## 2. Анализ данных 

### 2.1. Самые популярные товары

```python
top_products = df.groupby('product_name')['quantity'].sum().sort_values(ascending=False).head(5)
print(top_products)
```

**Вывод:** Позиции, которые покупают чаще всего по количеству штук. Важно держать их в наличии, чтобы избежать дефицита на складе.

### 2.2. Категории с наибольшей выручкой

```python
top_categories = df.groupby('category')['total_sales'].sum().sort_values(ascending=False)
print(top_categories)
```

**Вывод:** Товарные категории, приносящие основной доход. На них приходится главная доля продаж компании.

### 2.3. Месяцы с наибольшими продажами

```python
monthly_sales = df.set_index('order_date').resample('M')['total_sales'].sum().sort_values(ascending=False)
print(monthly_sales)
```

**Вывод:** Периоды пиковой активности покупателей. Позволяет увидеть сезонные колебания и понять, в какие месяцы продажи идут лучше всего.

### 2.4. Расчет среднего чека

```python
avg_check = df['total_sales'].mean()
print(avg_check)
```

**Вывод:** Показывает среднюю сумму одного заказа. Помогает оценить структуру покупок и сколько в среднем клиенты тратят за раз.

### 2.5. Клиенты с наибольшим количеством заказов

```python
top_customers = df['customer_id'].value_counts().head(5)
print(top_customers)
```

**Вывод:**Самые активные покупатели, которые оформляют заказы чаще остальных. На них стоит ориентироваться при внедрении программ лояльности.

## 3. Визуализация данных (5 обязательных графиков)

### Barplot — выручка по категориям товаров

```python
plt.figure(figsize=(8, 4))
df.groupby('category')['total_sales'].sum().plot(kind='bar', color='skyblue')
plt.title('Выручка по категориям товаров')
plt.xlabel('Категория')
plt.ylabel('Выручка')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Lineplot — динамика продаж по месяцам


```python
plt.figure(figsize=(10, 4))
df.set_index('order_date').resample('M')['total_sales'].sum().plot(kind='line', marker='o', color='b')
plt.title('Динамика продаж по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Выручка')
plt.tight_layout()
plt.show()
```

### Histogram — распределение стоимости заказов

```python
plt.figure(figsize=(8, 4))
df['total_sales'].plot(kind='hist', bins=30, color='green', edgecolor='black')
plt.title('Распределение стоимости заказов')
plt.xlabel('Сумма заказа')
plt.ylabel('Количество')
plt.tight_layout()
plt.show()
```

### Pie chart — распределение способов оплаты

```python
plt.figure(figsize=(6, 6))
df['payment_method'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Распределение способов оплаты')
plt.ylabel('')  # убираем лишнюю подпись оси y
plt.tight_layout()
plt.show()
```

### Barplot — топ клиентов по сумме покупок

```python
plt.figure(figsize=(8, 4))
df.groupby('customer_id')['total_sales'].sum().sort_values(ascending=False).head(5).plot(kind='bar', color='purple')
plt.title('Топ-5 клиентов по сумме покупок')
plt.xlabel('ID клиента')
plt.ylabel('Сумма покупок')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
